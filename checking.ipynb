```python
import pandas as pd
print(pd.__version__)

```

```python
df = pd.read_csv("collisions.csv", low_memory=False)
print("Shape:", df.shape)
```

```python
print("\nColumns:", df.columns.tolist())
```

```python
print("\nMissing % per column:")
print((df.isna().mean() * 100).round(1).sort_values(ascending=False))
```

```python
print("\nDuplicate COLLISION_ID:", df["COLLISION_ID"].duplicated().sum())
print("\nBorough counts:")
print(df["BOROUGH"].value_counts(dropna=False))
```

```python
dates = pd.to_datetime(df["CRASH DATE"], format="%m/%d/%Y", errors="coerce")
print("\nDate range:", dates.min(), "to", dates.max())
```

```python
print("Bad dates:", dates.isna().sum())

print("\nTop vehicle types:")
print(df["VEHICLE TYPE CODE 1"].value_counts().head(15))
```

```python


pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# dtype=str keeps everything exactly as in the file (raw view)
df = pd.read_csv("collisions.csv", dtype=str, low_memory=False)

print("Rows, Columns:", df.shape)
print("\n===== FIRST 10 ROWS =====")
print(df.head(10))

print("\n===== RANDOM 10 ROWS =====")
print(df.sample(10, random_state=1))

print("\n===== COLUMN-BY-COLUMN SUMMARY =====")
for col in df.columns:
    s = df[col]
    print("\n---", col, "---")
    print("Missing:", s.isna().sum(), f"({s.isna().mean()*100:.1f}%)")
    print("Unique values:", s.nunique())
    print("Sample values:", s.dropna().unique()[:8].tolist())
    print("Top 5 values:")
    print(s.value_counts(dropna=False).head(5))
```
